## 1. Metastore Level storage

- Metastore is attached to the Data Lake
- Catalog is Managed
- Schema is Managed
- Object is Managed

Now, data for our object[managed tables] will be saved in the Metastore Data Lake

```python
%sql
Create CATALOG IF NOT EXISTS managed_catalog;

CREATE SCHEMA IF NOT EXISTS managed_catalog.managed_schema;

CREATE TABLE IF NOT EXISTS managed_catalog.managed_schema.managed_table 
(
    id INT,
    name STRING
) ;
```

```python
%sql
INSERT INTO managed_catalog.managed_schema.managed_table
VALUES (1, 'John'), (2, 'Mary'), (3, 'Jane');
```

```python
%sql
SELECT * FROM managed_catalog.managed_schema.managed_table
```

## 2. Catalog Level storage

- Metastore is attached to the Data Lake
- Catalog is attached to the Data Lake
- Schema is Managed
- Object is Managed

Now, data for our object[managed tables] will be saved in the Catalog Data Lake

```python
%sql
CREATE CATALOG IF NOT EXISTS ext_catalog
MANAGED LOCATION 'abfss://metastore@devarmanunitycatalogdl.dfs.core.windows.net/catalog';

CREATE SCHEMA IF NOT EXISTS ext_catalog.managed_schema;

CREATE TABLE IF NOT EXISTS ext_catalog.managed_schema.managed_table 
(
    id INT,
    name STRING
) ;
```

```python
%sql
INSERT INTO ext_catalog.managed_schema.managed_table
VALUES (1, 'John'), (2, 'Mary'), (3, 'Jane');
```

```python
%sql
SELECT * FROM ext_catalog.managed_schema.managed_table
```

## External Tables

```python
%sql
CREATE CATALOG IF NOT EXISTS ext_catalog
MANAGED LOCATION 'abfss://metastore@devarmanunitycatalogdl.dfs.core.windows.net/catalog';

CREATE SCHEMA IF NOT EXISTS ext_catalog.ext_schema
MANAGED LOCATION 'abfss://metastore@devarmanunitycatalogdl.dfs.core.windows.net/schema';

CREATE TABLE IF NOT EXISTS ext_catalog.ext_schema.ext_table
(
    id INT,
    name STRING
) 
LOCATION 'abfss://raw@devarmanunitycatalogdl.dfs.core.windows.net/ext_table';
```

```python
%sql
INSERT INTO ext_catalog.ext_schema.ext_table
VALUES (1, 'Arman'), (2, 'KLS'), (3, 'Madhur');
```

```python
%sql
SELECT * FROM ext_catalog.ext_schema.ext_table
```

```python
%sql
INSERT OVERWRITE ext_catalog.ext_schema.ext_table
VALUES (1, 'Arman'), (2, 'KLS'), (3, 'Madhur');
```

```python
%sql
SELECT * FROM ext_catalog.ext_schema.ext_table
```

## Data Masking - Unity Catalog Functions

```python
%sql
CREATE TABLE ext_catalog.ext_schema.employee
(
    id INT,
    name STRING,
    salary INT
);

INSERT INTO ext_catalog.ext_schema.employee
VALUES (1, 'John', 100000),
       (2, 'Jane', 120000),
       (3, 'Bob', 80000),
       (4, 'Alice', 90000),
       (5, 'Mary', 110000);
    
SELECT * FROM ext_catalog.ext_schema.employee;
```

###Column Mask Examples -- Create a Function and apply it on table

```python
%sql
CREATE OR REPLACE FUNCTION ext_catalog.ext_schema.maskingk(salary STRING)
RETURN CASE WHEN is_account_group_member("admins") THEN salary ELSE '***_***_***' END
```

```python
%sql
ALTER TABLE ext_catalog.ext_schema.employee
ALTER COLUMN salary SET MASK ext_catalog.ext_schema.maskingk
```

```python
%sql
SELECT * FROM ext_catalog.ext_schema.employee
```
